```python
# encapsulation means the wrapping of data and function into a single units
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount

account = BankAccount("Ashim",1000)
account.deposit(500)

print(account.get_balance())
        
```

```python
#another encapsulation example
class Stock:
    def __init__(self, items, quantity):
        self.items = items
        self.__quantity = quantity

    def stock_aa(self):
        return self.__quantity

    def calculate(self, account):
        if account > 0:
            self.__quantity += account

s1 = Stock("Clothes",3)
s1.calculate(4)
print(s1.stock_aa()) 
```

```python
class Bank:
    def __init__(self,account_holder, balance=0):
        self.account_holder = account_holder
        self.__balance = balance
        
#deposit money
    def deposite(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Successfully depsosited amount")
        else:
            print("Depsite fail due to number less than zero")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Invalid, insufficient balance")
        elif amount <= 0:
            print("The number must be greater than zero")
        else:
            print("Amount withdrawn successfully")

    def get_balance(self):
        return self.__balance

    # Display account summary
    def display_account_info(self):
        print("=" * 30)
        print(f"       ACCOUNT SUMMARY")
        print("=" * 30)
        print(f"Holder Name     : {self.account_holder}")
        print(f"Current Balance : ${self.__balance:.2f}")
        print("=" * 30 + "\n")

acc = Bank("Ashim",10000)
acc.deposite(700)
acc.display_account_info()

```

```python
class Student:
    def __init__(self, name, marks):
        self.name=name
        self.__marks=marks

    def get_marks(self):
        return self.__marks

    def update_marks(self, new_marks):
        if 0 <= new_marks <= 100:
            print("Marks Updated Successfully!")
        else:
            print("Unable to Update Marks!")

    def display(self):
        print("Name:",self.name)
        print("Marks:",self.get_marks())


s1 = Student("Ashim",95)

s1.display()
s1.update_marks(98)
s1.display()
```

```python
class ATM:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance


    def check_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"The amount added is {amount} and total is {self.__balance }")
        else:
            print("Please enter a positive amount")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Invalid! Amount insufficient")
        else:
            print("Amount transferred successfully")

    def display(self):
        print("Name:",self.owner)
        print("Amount:",self.check_balance())
            

r1 = ATM("Ashim Acharya",40000)
r1.check_balance()
r1.display()
r1.deposit(4000)
```

```python
class OrderSupport:
    def __init__(self, order_id, status):
        self.order_id = order_id
        self.__status = status
        
    def get_status(self):
        return self.__status

    def update_status(self, new_status):
        self.__status = new_status
        print("Order status update successfully")

    def display(self):
        print("Order ID:",self.order_id)
        print("Order Status:",self.get_status())

#creating objects
c1 = OrderSupport(1,"Processing")
c1.display()
```
