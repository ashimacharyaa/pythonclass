```python
value={}
def word_frequency(sentence):
    for i in sentence.split():
        if i in value:
            value[i] += 1
        else:
            value[i] = 1
            
    return value

result = word_frequency("He is ram and ram is hero")
print(result)
```

```python
def second_largest(numbers):
    unique_numbers=[]

    for num in numbers:
        if num not in unique_numbers:
            unique_numbers.append(num)

    unique_numbers.sort()

    if len(unique_numbers) < 2:
        return None

    return unique_numbers[-2]

print(second_largest([10,20,40,40,30]))
```

```python
def word_count(word):
    words = word.split()
    w = {}

    for wo in words:
        if wo in w:
            w[wo] += 1
        else:
            w[wo] = 1
    return w

print(word_count("My name is ashim and my sis is aashika"))
        
```

```python
#idea behind unique number sort 
def third_largest(number):
    unique = []
    for i in number:
        if i not in unique:
            unique.append(i)
            
    unique.sort()

    return unique[-3]

print(third_largest([20,30,40,60]))
```

```python
def remove_duplicates(items):
    dup = []
    for word in items:
        if word not in dup:
            dup.append(word)

    return dup

print(remove_duplicates([1,2,2,4,6]))           
```

```python
def is_prime(number):

    if number <= 1:
        return False

    for i in range(2, int(number ** 0.5) + 1):

        if number % i == 0:
            return False

    return True

print(is_prime(20))
print(is_prime(17))
```

```python
def calculate_numbers(numbers):
    
    sorted_numbers = sorted(numbers)
    
    mean = sum(numbers)/len(numbers)

    n = len(sorted_numbers)

    if n % 2 == 0:
        median = (
            sorted_numbers[n // 2 - 1] +
            sorted_numbers[n // 2]
        ) / 2
    else:
        median = sorted_numbers[n // 2]
        
    minimum = min(numbers)
    maximum = max(numbers)

    return {
        "mean":mean,
        "median":median,
        "minimum":minimum,
        "maximum":maximum,
        "range":maximum-minimum
    }

print(calculate_numbers([10,20,30,40,50]))
```
